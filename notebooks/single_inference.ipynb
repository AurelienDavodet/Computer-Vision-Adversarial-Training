# Single inference

```python
import sys
import os

parent_dir = os.path.abspath(os.path.join(".."))
sys.path.append(parent_dir)
```

```python
import torch
from PIL import Image
import torchvision.transforms as transforms
from utils.cnn_architecture import SimpleCNN
from utils.data_transformation import transform
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

### Load the image

```python
img = Image.open('../data/cat.jpg')
img = transform(img)
img = img.unsqueeze(0)
img = img.to(device)
```

### Load the model

```python
# Basic model
# model = torch.load('../models/cifar10_basic_model.pth')

# Robust model
model = torch.load('../models/adversarially_trained_model.pth')
```

```python
model.to(device)
model.eval()
```

### Perform inference

```python
with torch.no_grad():
    outputs = model(img)

_, predicted_class = torch.max(outputs, 1)

classes = ['plane', 'car', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

print(f'Predicted class: {classes[predicted_class.item()]}')
```
